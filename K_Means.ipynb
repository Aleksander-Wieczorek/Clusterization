```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
```

```python
def import_data(file_path):
    data = np.loadtxt(file_path)
    return data
```

```python
max_iters=100
```

```python
def assign_labels(X,centroids):
    distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)
    return np.argmin(distances, axis=1)
```

```python
def fit(X,k):
    centroids = X[np.random.choice(len(X), size=k, replace=False)]
    for _ in range(max_iters):
        # Calculate distances and assign clusters

        closest_centroids = assign_labels(X, centroids)

        new_centroids = np.array([X[closest_centroids == i].mean(axis=0) for i in range(k)])
        if np.all(centroids == new_centroids):
            break
        centroids = new_centroids
    return centroids
```

```python
path = f'data_18D'

X_train = import_data(f'dane/rp.data')
X_train = np.array(X_train)
ks = range(1,50)
distances = []
for k in ks:
    centroids = fit(X_train, k)
    distance = np.sum(np.min(np.linalg.norm(X_train[:, np.newaxis] - centroids, axis=2), axis=1))
    distances.append(distance)
centroids = fit(X_train, 4)
plt.plot(ks, distances, marker='o')
plt.xlabel('Liczba klastrów (k)')
plt.ylabel('Suma odległości do najbliższego centroidu')
plt.title('Metoda łokcia dla wyboru liczby klastrów')
plt.savefig(f'kmeans_fig/{path}_elbow_method.png', dpi=300)
plt.show()

fig, ax = plt.subplots(figsize=(8, 6))

ax.scatter(
    X_train[:, 0],
    X_train[:, 1],
    c="#7F77DD",
    label="Punkty",
    s=80,
    edgecolors="white",
    linewidths=0.6,
    alpha=0.9,
)

ax.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker='x',
    s=200,
    c='black',
    linewidths=2.5,
    label='Centroidy',
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Wykres punktowy 2D")
ax.legend(title="Legenda")
ax.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()  
plt.savefig(f'kmeans_fig/{path}_kmeans.png', dpi=300)
plt.show()
```
