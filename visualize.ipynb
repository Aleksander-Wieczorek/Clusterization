```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = np.loadtxt("dane/dane_2D_7.txt", delimiter="\t")
from matplotlib.colors import ListedColormap

x      = data[:, 0]
y      = data[:, 1]
labels = data[:, 2].astype(int)


classes     = np.unique(labels)
colors      = ["#7F77DD", "#1D9E75", "#D85A30", "#D4537E", "#378ADD","#A78F27","#331313","#A1B2C3","#F0E68C","#8B0000","#2E8B57","#FF6347","#4682B4","#D2691E","#9ACD32","#FF4500","#6A5ACD","#20B2AA","#FF69B4","#8A2BE2","#5F9EA0","#FF1493","#B22222","#00CED1","#FF8C00","#7B68EE","#3CB371","#FF6347","#BA55D3","#00FA9A","#FF4500","#9370DB","#66CDAA","#FF69B4","#8A2BE2","#5F9EA0","#FF1493","#B22222","#00CED1","#FF8C00","#7B68EE","#3CB371","#FF6347","#BA55D3","#00FA9A","#FF4500","#9370DB","#66CDAA","#FF69B4","#8A2BE2","#5F9EA0","#FF1493","#B22222","#00CED1","#FF8C00","#7B68EE","#3CB371","#FF6347","#BA55D3","#00FA9A","#FF4500","#9370DB","#66CDAA"]
cmap        = ListedColormap(colors[:len(classes)])

fig, ax = plt.subplots(figsize=(8, 6))

for i, cls in enumerate(classes):
    mask = labels == cls
    ax.scatter(
        x[mask], y[mask],
        c=colors[i],
        label=f"Klasa {cls}",
        s=80,
        edgecolors="white",
        linewidths=0.6,
        alpha=0.9,
    )

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Wykres punktowy 2D")
ax.legend(title="Klasy")
ax.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.savefig("2D_7.png", dpi=150) 
plt.show()
```

```python

from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

data = np.loadtxt("dane/rp.data")
labels = data[:, -1].astype(int)
X = data[:, :-1] 
embedding = PCA(n_components=2).fit_transform(X)

fig, ax = plt.subplots(figsize=(10, 8))

scatter = ax.scatter(
    embedding[:, 0],
    embedding[:, 1],
    c=labels,
    cmap='tab20',
    s=35,
    alpha=0.8,
    edgecolors='white',
    linewidths=0.3
)

ax.set_xticks([])
ax.set_yticks([])

ax.set_title("PCA - Redukcja wymiarów")
ax.set_xlabel("PCA 1")
ax.set_ylabel("PCA 2")
legend1 = ax.legend(*scatter.legend_elements(), title="Klasy")
plt.tight_layout()
plt.show()
```
